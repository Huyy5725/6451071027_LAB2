```python
import pandas as pd


def calc_f_measure(precision, recall):
  if precision + recall == 0:
    return 0.0
  return 2 * (precision * recall) / (precision + recall)


print("Họ tên: Nguyễn Gia Huy - MSSV: 6451071027")
```

## Thống kê kết quả tính toán Precision & Recall từ các bài tập cơ bản


```python
print("Họ tên: Nguyễn Gia Huy - MSSV: 6451071027")
data_f = [
    {
        'Bài tập': 'Bài 1 (Đồ chơi - ID3)',
        'Precision': 0.75,
        'Recall': 1.00,
        'F-measure': calc_f_measure(0.75, 1.00),
    },
    {
        'Bài tập': 'Bài 1 (Đồ chơi - Naïve Bayes)',
        'Precision': 0.75,
        'Recall': 1.00,
        'F-measure': calc_f_measure(0.75, 1.00),
    },
    {
        'Bài tập': 'Bài 2 (Cảm xúc - CART)',
        'Precision': 0.50,
        'Recall': 0.50,
        'F-measure': calc_f_measure(0.50, 0.50),
    },
    {
        'Bài tập': 'Bài 2 (Cảm xúc - Naïve Bayes)',
        'Precision': 0.50,
        'Recall': 0.50,
        'F-measure': calc_f_measure(0.50, 0.50),
    },
]

df_fmeasure = pd.DataFrame(data_f)
```

##In kết quả

```python
print("Họ tên: Nguyễn Gia Huy - MSSV: 6451071027")
print("\n=== BẢNG TÍNH GIÁ TRỊ F-MEASURE CÁC BÀI TẬP CƠ BẢN ===")
print(df_fmeasure.to_string(index=False))
```
